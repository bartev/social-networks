### Question 1

Using networkx, load up the directed multigraph from `assets/email_network.txt`. Make sure the node names are strings.

*This function should return a directed multigraph networkx graph.*

```python
def answer_one():
    file = "assets/email_network.txt"
    G = nx.MultiDiGraph()

    with open(file) as f:
        next(f) # Skip header row
        for line in f:
            sender, recip, ts = line.strip().split()
            G.add_edge(sender, recip, time=int(ts))
    return G
```

### Question 11

Which node in G_sc has the most shortest paths to other nodes whose distance equal the diameter of G_sc?


For the node with the most such shortest paths, how many of these paths are there?


*This function should return a tuple (name of node, number of paths).*

```python
def answer_eleven():
    from collections import defaultdict

    G_sc = answer_six()
    diameter = nx.diameter(G_sc)
    periph = nx.periphery(G_sc)
    counts = defaultdict(int)

    for node in periph:
        # nodes whose path length to a periphery node is diameter away
        source_nodes = {
            source: distance
            for source, distance in dict(
                nx.single_target_shortest_path_length(G_sc, node)
            ).items()
            if distance == diameter
        }
        for source in source_nodes:
            counts[source] += 1
    top_count = max(counts.values())
    top_nodes = [node for node, count in counts.items() if count == top_count]
    return top_nodes[0], top_count
```

### Question 12

Suppose you want to prevent communication flow from the node that you found in question 11 to node 10. What is the smallest number of nodes you would need to remove from the graph (you're not allowed to remove the node from the previous question or 10)? 

This is the `node_connectivity`

`node_cuts` would give us which nodes to remove.

*This function should return an integer.*

```python
def answer_twelve():
    G_sc = answer_six()
    start_node = answer_eleven()[0]
    end_node = '10'
    connectivity = nx.node_connectivity(G_sc, start_node, end_node)
    node_cuts = nx.minimum_node_cut(G_sc, start_node, end_node)
    print(f"{connectivity=}")
    print(f"{node_cuts}")
    return connectivity
```
